# SQL in Python

```python
import sqlite3
import pandas as pd
```

```python
# Create database in sql 

conn = sqlite3.connect('test_database.sqlite')

#table details
conn.execute("CREATE TABLE users (first_name TEXT, country TEXT, budget INTEGER)")

# commit and save
conn.commit()
```

```python
# insert data 

cursor = conn.cursor()

# write sql query to insert data in sql table
cursor.execute(
    """
        INSERT INTO users VALUES 
        ('Abhishek', 'India', 1000000),
        ('Madhav', 'India', 5000000),
        ('Manish', 'US', 10000),
        ('Rohit', 'Russia', 900000)
    """
)

# commit and save
conn.commit()

print("Data Inserted Succesfully")
```

# check inserted data into table

```python
conn = sqlite3.connect('test_database.sqlite')

# query 
query = """ SELECT * FROM users"""

df_result = pd.read_sql(query ,conn)

df_result
```

```python
# aggregation 
query = """
        SELECT country, sum(budget) as Total_budget
        FROM users
        GROUP BY country;
"""
df_agg = pd.read_sql(query ,conn)

df_agg
```
